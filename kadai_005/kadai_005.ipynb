# 課題005 Matplotlibで、可視化した複数のグラフを1つの描画キャンパス上で同時に表示しよう

`plt.subplots(nrows, ncols)` で 2行3列＝6個の領域をまとめて作り、
そのうち4つに「折れ線グラフ」「sin関数」「ヒストグラム」「散布図」を描き分けます。

- Figure … 描画領域全体（キャンバス）
- Axes  … ひとつひとつのグラフを描く領域。`axes[行, 列]` で指定する

領域どうしが重ならないよう、`tight_layout=True` でレイアウトを自動調整しています。
残りの2領域は課題の指示どおり、何も描かず空のまま残しています。

```python
# Step1. matplotlib.pyplot と numpy をインポートする
import numpy as np
import matplotlib.pyplot as plt

# 乱数を使うグラフがあるので、実行するたびに結果が変わらないよう種を固定する
rng = np.random.default_rng(0)

# Step2. 6個のグラフを表示する領域を作成する
# nrows=2, ncols=3 で 2行3列=6領域のサブプロットになる
# tight_layout=True で各領域が重ならないようレイアウトを自動調整する
fig, axes = plt.subplots(nrows=2, ncols=3, tight_layout=True, figsize=(10, 6))

# Step3. 各領域にグラフを描く。axes[行, 列] で描き込む領域を指定する
# 3-1. 折れ線グラフ (0行0列)
y0 = [1, 2, -5, 2]
axes[0, 0].plot(y0)
axes[0, 0].set_title("line")

# 3-2. sin関数 (0行1列)
x1 = np.arange(0, 2 * np.pi, 0.1)
y1 = np.sin(x1)
axes[0, 1].plot(x1, y1)
axes[0, 1].set_title("sin")

# 3-3. ヒストグラム (0行2列) 正規分布に従う1000個の乱数の分布を見る
y2 = rng.normal(loc=50, scale=10, size=1000)
axes[0, 2].hist(y2, bins=20)
axes[0, 2].set_title("histogram")

# 3-4. 散布図 (1行0列)
x3 = rng.random(100)
y3 = rng.random(100)
axes[1, 0].scatter(x3, y3)
axes[1, 0].set_title("scatter")

# 残りの2領域 axes[1, 1] と axes[1, 2] は、課題の指示どおり何も描かずに残す

# Step4. グラフを表示する
plt.show()
```
